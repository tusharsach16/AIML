```python
import pandas as pd
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.DataFrame({
    "Hours": [2, 5, 8],
    "CGPA": [6.0, 7.5, 9.0],
    "Salary": [4, 8, 12],
    "Placement": ["No", "Yes", "Yes"]
})


X = df[["Hours", "CGPA", "Salary"]]
Y = df["Placement"]

model = DecisionTreeClassifier(
    criterion = "gini",
    random_state = 42
)

model.fit(X, Y)

print("Model Trained")
```

```python
new_student = pd.DataFrame({
    "Hours": [6],
    "CGPA": [8.0],
    "Salary": [9]
})

prediction = model.predict(new_student)
print("Predicted Placement: ", prediction[0])
```
