```python
import time
```

```python
data = np.random.randn(1000000)
data
```

```python
start = time.time()
result = [x**2 for x in data]
loop_time = time.time() - start
print(f"Loop time: {loop_time:.4f} seconds")
```

```python
start = time.time()
result = data **2
vectorised_time = time.time() - start
print(f"Vectorised time: {vectorised_time:.4f} seconds")
```

```python
print(f"NumPy is {loop_time/vectorised_time:.1f}x faster!")
```
